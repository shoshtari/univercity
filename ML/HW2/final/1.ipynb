```python
import numpy as np 
import math
import matplotlib.pyplot as plt
from typing import Optional, Tuple
```

```python
def parzen_mean(x: float, h: float, a: float) -> float:
    if x < 0:
        return 0
    if x < a:
        return (1 - math.e**(-x / h) ) / a
    return math.e**(-x / h) * (math.e**(a/h) - 1) / a
def plot_parzen(h: float, a: float, plot_range: Optional[Tuple[float, float]] = None):
    if plot_range is None:
        plot_range = (0, 1.5 * a)
    x = np.linspace(plot_range[0], plot_range[1], 1000)
    y = [parzen_mean(i, h, a) for i in x]
    plt.plot(x, y, label=f"{h = }")
plt.figure(figsize=(15, 9))
plot_parzen(h = 1, a = 1)
plot_parzen(h = 0.25, a = 1)
plot_parzen(h = 0.0625, a = 1)
plt.axvline(1, lw=5, alpha=0.3)
plt.legend(fontsize='xx-large')
```
