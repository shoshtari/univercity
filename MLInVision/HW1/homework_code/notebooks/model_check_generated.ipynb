We try to run a simple model on some generated data to see if it works correctly.

```python
from models.model import ANN, Layer
import matplotlib.pyplot as plt
import numpy as np
from models.layer.initializer import Initializer
from models.layer.activation import Relu, Softmax
```

```python
def make_moons(n_samples=500, noise=0.1):
    n = n_samples // 2

    theta = np.linspace(0, np.pi, n)
    x1 = np.c_[np.cos(theta), np.sin(theta)]
    x2 = np.c_[1 - np.cos(theta), -np.sin(theta)]

    X = np.vstack([x1, x2])
    X += noise * np.random.randn(*X.shape)

    y = np.vstack([np.zeros((n, 1)), np.ones((n, 1))])
    y = y.reshape(len(y))

    return X, y


X, y = make_moons(n_samples=2000)
print(X.shape)
print(y.shape)
```

```python
plt.figure(figsize=(4, 3))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="berlin", alpha=0.5)
```

```python
model = ANN(
    layers=[
        Layer(
            layer_size=64,
            previous_layer_size=2,
            activation=Relu(),
            W_initializer=Initializer.he,
            b_initializer=Initializer.constant(0.01),
        ),
        Layer(
            layer_size=4,
            previous_layer_size=64,
            activation=Relu(),
            W_initializer=Initializer.he,
            b_initializer=Initializer.constant(0.01),
        ),
        Layer(
            layer_size=2,
            previous_layer_size=4,
            activation=Softmax(),
            W_initializer=Initializer.xavier,
            b_initializer=Initializer.constant(-0.01),
        ),
    ],
    loss="cross_entropy",
    learning_rate=0.005,
    random_state=42,
)
y_train = np.eye(2)[y.reshape(len(y)).astype(int)]
print(model.evaluate(X, y))
model.fit(X=X, Y=y_train, epochs=100, logging_step=0)
print(model.evaluate(X, y_train))
```
