```python
def count_class(Y: NDArray[np.number], normalize: bool = False):
    Y = np.argmax(Y, axis=1)
    counts = np.unique_counts(Y)
    assert (counts.values == range(10)).all()
    ans = counts.counts
    if normalize:
        ans = ans / ans.sum()
    return ans


count_class(data.y_train)
```

```python
classes = np.arange(10)

train = count_class(data.y_train, normalize=True)
val = count_class(data.y_val, normalize=True)
test = count_class(data.y_test, normalize=True)

bar_width = 0.25

plt.figure(figsize=(15, 6))
plt.bar(classes - bar_width, train, width=bar_width, label="train")
plt.bar(classes, val, width=bar_width, label="val")
plt.bar(classes + bar_width, test, width=bar_width, label="test")

plt.xlabel("Class")
plt.ylabel("Accuracy")
plt.title("Classes distribution")
plt.xticks(classes)
plt.legend(loc="upper right")
plt.show()

table = Table()
table.add_column("name")
table.add_column("std")
table.add_row("train", str(round(train.std(), 6)))
table.add_row("val", str(round(val.std(), 6)))
table.add_row("test", str(round(test.std(), 6)))
console = Console()
console.log(table)
```
