```python
# for report
row = ds_train[0]
table = []
row = {
    "tokens": row[0],
    "pos_tags": row[1]["pos_tags"],
    "chunk_tags": row[1]["chunk_tags"],
    "ner_tags": row[1]["ner_tags"],
}
assert len(row["tokens"]) == len(row["pos_tags"])
assert len(row["tokens"]) == len(row["chunk_tags"])
assert len(row["tokens"]) == len(row["ner_tags"])

n = len(row["tokens"])
table = [{}, {}, {}]
for i in range(n):
    token = row["tokens"][i]
    table[0][token] = row["pos_tags"][i]
    table[1][token] = row["chunk_tags"][i]
    table[2][token] = row["ner_tags"][i]

df = pd.DataFrame(table)
print(df.to_latex().replace("_", r"\_"))
```
