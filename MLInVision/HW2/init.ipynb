```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def initialization(student_seed):
    np.random.seed(student_seed)

    input1 = np.random.choice(np.arange(-2,3), size=(3,3),  p=[0.1, 0.2, 0.4, 0.2, 0.1])
    input2 = np.random.choice(np.arange(-2,3), size=(3,3),  p=[0.05, 0.15, 0.6, 0.15, 0.05])
    conv1 = np.random.choice(np.arange(-2,3), size=(2,2),  p=[0.2, 0.25, 0.1, 0.25, 0.2])
    conv2 = np.random.choice(np.arange(-2,3), size=(2,2),  p=[0.15, 0.25, 0.2, 0.25, 0.15])
    lin_weights= np.random.choice(np.arange(-4,5), size=4).reshape(4,1)
    biases = np.random.choice(np.arange(-1,2), size=2).reshape(1,2)

    fig, [[input1_ax, conv1_ax, lin_weights_ax], [input2_ax, conv2_ax, biases_ax]] = plt.subplots(2, 3, figsize=(16,8))

    input1_ax = sns.heatmap(input1,annot=True,cmap=['black'],cbar=False,xticklabels=False,yticklabels=False,ax=input1_ax)
    input2_ax = sns.heatmap(input2,annot=True,cmap=['black'],cbar=False,xticklabels=False,yticklabels=False,ax=input2_ax)
    conv1_ax = sns.heatmap(conv1,annot=True,cmap=['black'],cbar=False,xticklabels=False,yticklabels=False,ax=conv1_ax)
    conv2_ax = sns.heatmap(conv2,annot=True,cmap=['black'],cbar=False,xticklabels=False,yticklabels=False,ax=conv2_ax)
    lin_weights_ax = sns.heatmap(lin_weights,annot=True,cmap=['black'],cbar=False,xticklabels=False,yticklabels=['w1','w2','w3','w4'],ax=lin_weights_ax)
    biases_ax = sns.heatmap(biases,annot=True,cmap=['black'],cbar=False,xticklabels=['convoloution layer','linear layer'],yticklabels=False,ax=biases_ax)

    input1_ax.set_title('raw input (channel1)')
    input2_ax.set_title('raw input (channel2)')
    conv1_ax.set_title('conv1 weights')
    conv2_ax.set_title('conv2 weights')
    lin_weights_ax.set_title('linear layer weights')
    biases_ax.set_title('biases')

    plt.show();
```

```python
'Write the last 3 digits of your student number here (as Integer):'
initialization(256)
```
