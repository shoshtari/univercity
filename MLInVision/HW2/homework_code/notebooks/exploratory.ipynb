```python

import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter, ScalarFormatter
import numpy as np

from data.data_loader import get_dataloader, MNISTDataset
```

```python
dataset = MNISTDataset('train')
```

```python
counts = np.unique_counts(dataset._y)

color = plt.cm.viridis_r(counts.counts / max(counts.counts))


fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

ax1.set_ylabel('Count')
ax2.set_ylabel('Percentage')
ax2.yaxis.set_major_formatter(PercentFormatter(len(dataset)))

ax1.bar(counts.values, counts.counts, color=color)
ax2.bar(counts.values, counts.counts, color=color)


plt.tight_layout()
```
