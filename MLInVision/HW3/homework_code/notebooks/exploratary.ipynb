All alpha channel pixels has same data

```python
def counts_dataset(ds):
    n = len(ds.class_names)
    counts = np.zeros(n, dtype=np.int64)
    for i in tqdm(range(len(ds)), desc="Processing dataset..."):
        mask = ds[i][1].numpy()
        for c in range(n):
            counts[c] += int((mask == c).sum())
    return counts


counts1 = counts_dataset(ds1)
counts2 = counts_dataset(ds2)

# Plot side-by-side bar charts
plt.figure(figsize=(16, 6))
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

plt.subplot(1, 2, 1)
plt.bar(range(len(ds1.class_names)), counts1)
plt.xticks(range(len(ds1.class_names)))
plt.title("Dataset1: Pixel counts per label")

plt.subplot(1, 2, 2)
plt.bar(range(len(ds2.class_names)), counts2)
plt.xticks(range(len(ds2.class_names)))
plt.xticks(rotation=45, ha="right")
plt.title("Dataset2: Pixel counts per label")

plt.tight_layout()
plt.show()
```
