```python
from scipy.stats import chi2, norm, t
import numpy as np
```

```python
def ci_variance(s2: float, n: int, alpha=0.95):

    s2 = (n / (n - 1)) * s2 

    df = n - 1
    chi2_lower = chi2.ppf((1 + alpha)/2, df)
    chi2_upper = chi2.ppf((1 - alpha)/2, df)
    
    ci_lower = (n-1) * s2 / chi2_lower
    ci_upper = (n-1) * s2 / chi2_upper
    
    return ci_lower, ci_upper
print(ci_variance(49, 36))
```

```python
def ci_mean_t(n: int, s2: float, mean: float, alpha: float = 0.95):
    
    s2 = (n / (n - 1)) * s2 
    s = s2 ** (0.5)

    df = n - 1
    t_val = t.ppf((1 + alpha)/2, df)
    
    ci_lower = mean - (s / (n ** 0.5)) * t_val
    ci_upper = mean + (s / (n ** 0.5)) * t_val
    
    return ci_lower, ci_upper
print(28.4 - 1.18 * t.ppf(0.975, 35))
print(28.4 + 1.18 * t.ppf(0.975, 35))
print(ci_mean_t(36, 49, 28.4))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, t

def bootstrap_ci_basic(data, stat_func=np.mean, alpha=0.05, n_boot=10000):
    n = len(data)
    bootstrap_stats = []
    
    for _ in range(n_boot):
        bootstrap_sample = np.random.choice(data, size=n, replace=True)
        stat = stat_func(bootstrap_sample)
        bootstrap_stats.append(stat)
    
    bootstrap_stats = np.array(bootstrap_stats)
    
    ci_lower = np.percentile(bootstrap_stats, 100 * alpha/2)
    ci_upper = np.percentile(bootstrap_stats, 100 * (1 - alpha/2))
    
    return ci_lower, ci_upper, bootstrap_stats

# Example with your data
data = np.array([116.3, 112.6, 107.4, 88.4, 143.6, 113.6, 85.6, 97.2, 
                 121.3, 114.8, 93.3, 116.1, 93.8, 93.6, 98.3])

ci_mean, ci_upper, boot_means = bootstrap_ci_basic(data, stat_func=np.mean, n_boot=5000)
print(f"Bootstrap 95% CI for mean: [{ci_mean:.4f}, {ci_upper:.4f}]")
print(f"Sample mean: {np.mean(data):.4f}")
```
