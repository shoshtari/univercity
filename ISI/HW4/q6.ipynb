```python

def fact(n): 
    if n == 0 or n == 1: 
        return 1
    return n * fact(n - 1)
def C(n, r):
    return fact(n) // (fact(r) * fact(n - r))

def P(x):
    p = 0.7
    return C(100, x) * (p ** x) * ((1 - p) ** (100 - x))
print(1 - P(69) - P(70) - P(71))
```

```python
n = 100
p = 0.7
mean = n * p
std_dev = (n * p * (1 - p)) ** 0.5
print("Mean:", mean)
print("Standard Deviation:", std_dev)
```
