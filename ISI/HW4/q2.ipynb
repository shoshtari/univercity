```python
import numpy as np
from scipy import stats
```

```python
raw_data = '''1 4.62 4.54
2 4.58 4.49
3 4.60 4.53
4 4.65 4.57
5 4.59 4.50
6 4.63 4.56
7 4.61 4.64
8 4.56 4.47
9 4.64 4.57
10 4.60 4.51
11 4.57 4.59
12 4.62 4.53
13 4.58 4.49
14 4.63 4.66
15 4.59 4.52'''.strip().split('\n')
```

```python
tmp = [list(map(float, line.split()[1:])) for line in raw_data]
data = [i[1] - i[0] for i in tmp]
print("mean", np.mean(data))
print("std", np.std(data, ddof=1))
t = np.mean(data) / ((np.std(data, ddof=1) / np.sqrt(len(data))))
print(t)
```

```python
p_val = stats.t.cdf(t, df=len(data) - 1)
print("p-value", p_val)
```

```python
ME = stats.t.ppf(0.95, df=len(data) - 1) * (np.std(data, ddof=1) / np.sqrt(len(data)))
mean = np.mean(data)
print("95% CI:", (mean - ME, mean + ME))
print(np.exp(mean - ME), np.exp(mean + ME))
```
