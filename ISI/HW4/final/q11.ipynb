```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def fact(n): 
    if n == 0 or n == 1: 
        return 1
    return n * fact(n - 1)
def C(n, r):
    return fact(n) // (fact(r) * fact(n - r))

s = 0
for i in range(60):
    s += C(100, i) * (0.7 ** i) * (0.3 ** (100 - i))
for i in range(81, 101):
    s += C(100, i) * (0.7 ** i) * (0.3 ** (100 - i))
print(s)
```

```python
def power(p):
    s = 0
    for i in range(60):
        s += C(100, i) * (p ** i) * ((1 - p) ** (100 - i))
    for i in range(81, 101):
        s += C(100, i) * (p ** i) * ((1 - p) ** (100 - i))
    return s
p = np.linspace(0, 1, 100)
power_values = [power(pi) for pi in p]
plt.plot(p, power_values)
plt.xlabel('p')
plt.ylabel('Power')
plt.title('Power Function')
# line at p = 0.7
plt.axvline(x=0.7, color='r', linestyle='--', label='p = 0.7')
plt.legend()
```
