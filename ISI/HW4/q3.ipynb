```python
import numpy as np
from scipy import stats
```

```python
raw_data_a = """2.3 3.1 2.4 2.0
2.0 2.1 2.2 2.6
1.9 2.45 2.1 2.0
3.1 3.7 2.9 2.0
2.2 2.54 2.2 2.4
2.3 3.72 2.4 3.18
2.8 4.54 2.7 3.0
1.9 1.61 1.9 2.54
1.1 1.63 1.3 1.72""".strip().split('\n')
raw_data_b = """1.9 1.45 1.9 1.91
2.3 2.45 2.4 2.54
2.0 1.81 2.0 1.45
1.6 1.72 1.5 1.45
1.6 1.63 1.5 1.54
2.6 2.45 2.7 1.54
1.7 2.18 1.7 1.54""".strip().split('\n')
```

```python
tmp = [list(map(float, line.split())) for line in raw_data_a]
drug_a = [i[1] - i[0] for i in tmp]
placebo_a = [i[3] - i[2] for i in tmp]
tmp = [list(map(float, line.split())) for line in raw_data_b]
drug_b = [i[1] - i[0] for i in tmp]
placebo_b = [i[3] - i[2] for i in tmp]
```

```python
def t_test(sample_mean, sample_std, n):
    t = (sample_mean) / (sample_std / np.sqrt(n))
    dof = n - 1
    p_val = 2 * (1 - stats.t.cdf(abs(t), df=dof))
    
    return p_val
diff_a = np.array(drug_a) - np.array(placebo_a)
diff_b = np.array(drug_b) - np.array(placebo_b)
print("Drug A p-value:", t_test(
    np.mean(diff_a),
    np.std(diff_a, ddof=1), 
    len(diff_a), 
    0)
)
print("Drug B p-value:", t_test(
    np.mean(diff_b),
    np.std(diff_b, ddof=1), 
    len(diff_b), 
    0)
)
```
