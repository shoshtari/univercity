```python
from scipy import stats
import numpy as np
```

```python
O = np.array([
    26,
    32,
    40,
    30
])
E = np.array([
    16,
    48,
    48,
    16
])

```

```python
t = np.sum((O - E) ** 2 / E)
p_value = 1 - stats.chi2.cdf(t, df=3)
print("Chi-squared test statistic:", t)
print("p-value:", p_value)
```
