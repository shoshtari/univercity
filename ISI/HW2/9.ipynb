```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
COUNT = 1000
np.random.seed(42)
```

```python
normal_samples = np.random.standard_normal(size = (COUNT, 1))
normal_sample_mean = []
for i in range(COUNT):
    normal_sample_mean.append(np.mean(normal_samples[:i + 1]))
```

```python
cauchy_samples = np.random.standard_cauchy(size = (COUNT, 1))
cauchy_sample_mean = []
for i in range(COUNT):
    cauchy_sample_mean.append(np.mean(cauchy_samples[:i + 1]))
```

```python

plt.plot(normal_sample_mean, label = "normal", c="blue", lw=2)
plt.plot(cauchy_sample_mean, label="cauchy", c ="red", lw=2)
plt.axhline(0, lw=4, alpha=0.5)
plt.legend()
```
