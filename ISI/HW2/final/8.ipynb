```python
from typing import List
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
def print_float(*args: List[float]) -> None:
    for arg in args:
        if not isinstance(arg, float):
            raise TypeError("arg must be of type 'float'")
        print(f"{arg:.5f}", end = ' ')
    print()
```

```python
sigmas = (
    0.05,
    0.06,
    0.07
)
Zs = (
    1.96,
    2.57
)
for sigma in sigmas:
    for Z in Zs:
        width = Z * sigma / 8
        l = 12.08 - width 
        r = 12.08 + width 
        print_float(sigma, Z, width, l, r)
```

```python

def plot_t_and_z(sample_sizes: List[int]):
    t_vals = [stats.t.ppf(0.975, n-1) for n in sample_sizes]
    z_val = [1.96 for _ in sample_sizes]

    plt.plot(sample_sizes, t_vals, lw=2, label="T")
    plt.plot(sample_sizes, z_val, lw=2, label="Z")
    plt.xlabel('Sample Size (n)')
    plt.title('T and Z for alpha 95')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.legend()
plt.figure(figsize=(8, 5))
plt.subplot(1, 2, 1)
plot_t_and_z(list(range(2, 101)))
plt.subplot(1, 2, 2)
plot_t_and_z(list(range(40, 500)))
plt.show()
```
