```python
from typing import List
```

```python

groups = (
    (5_000, 80_000),
    (25_000, 30_000),
    (20_000, 15_000)
)

sampling_quota = (
    (333, 333, 333),
    (100, 500, 400),
    (276, 517, 207)
)

```

```python
def neyman():
    total = 0
    for i in range(3):
        total += groups[i][0] * groups[i][1]
    ans = []
    for i in range(3):
        numinator = groups[i][0] * groups[i][1]
        ans.append(numinator * 1000 / total)
    return ans
print(neyman())
```

```python
def calc_se(n: List[int]):
    ans = 0
    for i in range(3):
        w, sigma = groups[i]
        ans += w**2 * (sigma**2 / n[i])
    return ans ** 0.5
for quota in sampling_quota:
    print(f"{calc_se(quota):,}")
```
